```python
%sql
create catalog if not exists aws_databricks;
create schema if not exists aws_databricks.raw;
```

```python
import random
from pyspark.sql import Row
from datetime import datetime,timedelta
```

```python
help(datetime)
```

```python
base_time= datetime(2025,1,10,9,0,0)
batch_1 = [
    Row(
        event_id = i,
        event_type = random.choice(["click","view","purchase"]),
        user_id = random.randint(100,199),
        event_ts = base_time + timedelta(seconds= i * 30)
    )
    for i in range(1,11)
]

df = spark.createDataFrame(batch_1)
df.display()
```

```python
df.write.mode("append").saveAsTable("aws_databricks.raw.events")
```

```python
df = spark.sql('describe detail aws_databricks.raw.events') \
          .select("id") \
          .collect()[0][0]

print(f"location is {df}")
```
