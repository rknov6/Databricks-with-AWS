```python
from pyspark.sql import Row
```

```python
source_data = [
    Row(
        product_id = 1,
        name = 'Laptop',
        category = 'Electronics',
        price = 75000.00
    ),
    Row(
        product_id = 2,
        name = 'Mouse',
        category = 'Electronics',
        price = 5000.00
    ),
    Row(
        product_id = 3,
        name = 'Desktop',
        category = 'Electronics',
        price = 50000.00
    ),
    Row(
        product_id = 4,
        name = 'Chair',
        category = 'Furniture',
        price = 2000.00
    ),
    Row(
        product_id = 5,
        name = 'Table',
        category = 'Furniture',
        price = 7000.00
    )
]
```

```python
df = spark.createDataFrame(source_data,"product_id int, name string, category string, price double")
df.display()
```

```python
df.write.mode("append").saveAsTable("aws_databricks.raw.orders")
```

```python
new_data = [
    Row(product_id = 6,name = "Monitor",category = "Electronics",price = "65000.90")
]
new_df= spark.createDataFrame(new_data)
new_df.write.mode("append").saveAsTable("aws_databricks.raw.orders")
```

```python
new_column_data = [
    Row(product_id = 7,name = "Keyboard",category = "Electronics",price = "6000.90",discount_rate = 0.05)
]
new_column_df = spark.createDataFrame(new_column_data)
new_column_df.write.mode("append").option("mergeSchema","true").saveAsTable("aws_databricks.raw.orders")
```
