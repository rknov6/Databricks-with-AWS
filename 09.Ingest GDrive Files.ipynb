```python
schema_location = "/Volumes/dev/dbx_course/landing/_schemas/taxi"
checkpoint_location = '/Volumes/dev/dbx_course/landing/_checkpoints/taxi'

taxi_raw = (
    spark.readStream
         .format("cloudFiles")
         .option("databricks.connection","mydrive")
         .option("cloudFiles.format","csv")
         .option("cloudFiles.schemaLocation",schema_location)
         .option("header",True)
         .load("https://drive.google.com/drive/folders/10h0aHdoRiUx8QlbAZHUlD22avzzmdl5B")
)

(
    taxi_raw.writeStream
            .format("delta")
            .option("checkpointLocation",checkpoint_location)
            .trigger(availableNow=True)
            .table("dev.default.taxi")
            .awaitTermination()
)

spark.read.table("dev.default.taxi").display()
```
